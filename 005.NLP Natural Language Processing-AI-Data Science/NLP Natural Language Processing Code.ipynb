# Text Cleaning

Remove  punctuations

```python
import re
#Regular Expresssion library imported
corpus=[]

for i in range(len(sentences)):
    rp=re.sub('[^a-zA-Z]'," ",sentences[i])
    corpus.append(rp)
    
print(corpus)    
```

List of stop words which has no importances.

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

Remove Stop words

```python
corpus=[]
for i in range(len(sentences)):
    rp=re.sub('[^a-zA-Z]'," ",sentences[i]) #Remove punctuation HE is a good boy
    rp=rp.lower()#he is  a good boy
    rp=rp.split()#'he','is','a','good','boy'
    rp=[word for word in rp if not word in set(stopwords.words('english'))]#List comprehension
    rp=" ".join(rp)#'good boy'
    corpus.append(rp)
    
print(corpus)  
```

# Text Cleaning(Remove Punctuation+Remove Stopwords+Stemming/Lemmatization)

```python
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps=PorterStemmer()

corpus=[]
for i in range(len(sentences)):
    rp=re.sub('[^a-zA-Z]'," ",sentences[i])
    rp=rp.lower()
    rp=rp.split()
    rp=[ps.stem(word) for word in rp if not word in set(stopwords.words('english'))]
    rp=" ".join(rp)
    corpus.append(rp)
    
print(corpus)
```

# Vectorization

Count Vectorizer(Bag of Words) it count repetation of words

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
bow = cv.fit_transform(corpus).toarray()
bow
```

TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf=TfidfVectorizer()
tfidf=tf.fit_transform(corpus).toarray()
tfidf
```
