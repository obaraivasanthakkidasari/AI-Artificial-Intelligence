Sum of even numbers in between given range (Both Inclusive)

```python
lb = int(input(" Enter lower bound values"))
ub = int(input("Enter Upper bound values"))

s=0

for i in range(lb,ub+1):
    if i%2==0:
        s=s+i
        print(i, end = " ")
        
print('\nsum is', s)
     
```

Write a program which can compute the factorial of a given number

```python
a = int(input())

fact = 1

for i in range(a, 0, -1):
    fact = fact*i
    
print("Factorial of number", a,"is", fact)
```

# Loop Termination

# BREAK

```python
numbers = [1,2,3,4]

for i in numbers:
    if i ==3:
        break
print(i)
```

# Continue

```python
numbers = [1,2,3,4,5]

for num in numbers:
    if num %2 ==0:
        print(num)
        continue
```

# For loop with else

```python
numbers = [1,2,3,4]

for item in numbers:
    print(item)
    if item == 5:
        break
    
else:
    print("no item left in the list")
```
