# Modules

# 1.Python Math Module

```python
# Import the math module
import math

# Square root
print("Square root of 4:", math.sqrt(4))

# Exponential
print("e^1:", math.exp(1))

# Value of pi
print("Pi:", math.pi)

# Tangent
print("tan(1):", math.tan(1))
```

# 2.Import specific functions

```python
from math import sqrt, exp

print("Square root:", sqrt(4))
print("Exponential:", exp(1))
```

# 4. Python random module

```python
import random

# Generate a random decimal number between 0 and 1
print("Random number:", random.random())

# Generate a random integer between 1 and 40
print("Random integer:", random.randint(1, 40))

# Generate a random number from 0 to 100 with step 5
print("Random range:", random.randrange(0, 101, 5))
```

# 5. random.choice()

```python
import random

names = ["Sunny", "Bunny", "Chinny", "Vinny"]

selected_name = random.choice(names)

print("Randomly selected name:", selected_name)
```
