# 4. User Input and Basic Math 

```python
num1_str = input("Enter a number: ")        # Accepting num1_str from user.
num2_str = input("Enter another number: ")  #  # Accepting num2_str from user.
num1 = int(num1_str)                        # Converting string into intiger
num2 = int(num2_str)                        # Converting string into intiger
print(f"{num1} + {num2} = {num1 + num2}")   # Formatted print add num1 and num2 and add them. 
print(f"{num1} * {num2} = {num1 * num2}")   # Formatted print Multiply num1 and num2 and Multiply them. 
```

# 6. Lists: Modifying with Functions 

```python
data_points = [10, 20, 40, 50]            # Creating list with 4 items.
data_points.append(60)                    # Append: add item at the end of the list
print(f"After append: {data_points}")     # Print list after append using F-String print
data_points.insert(2, 30)                 # Insert at index 2 add item 30
print(f"After insert: {data_points}")     # Print list after Insert using F-String print
data_points.remove(40)                    # Remove item 40 from list
print(f"After remove: {data_points}")     # Print list after Insert using F-String print
```

# 10. Conditional Logic: if/else

```python
age = 20                                      # Declaring age variable
if age >= 18:                                 # Checking age is greater than equal to 18 or not By if condition
    print("Eligible to vote.")                # If Condition is True Then print Eligible to vote 
else:                                         # Else: If Condition is Flase
    print("Not eligible to vote.")            # Then print Not Eligible to vote 
```

# 11.Conditional Logic: if/elif/else 

```python
                                           # Conditional Logic: if/elif/else- Writing Multiple Conditions
temperature = 25                           # Declaring temperature variable
if temperature > 30:                       # Checking temperature is greater than to 30 or not By if condition
    print("It's hot outside!")             # If Condition is True Then print  It's hot outside!
elif temperature > 20:                     # Elif: If Condition is Flase then checking elif condition
                                           # Checking temperature is greater than 20 or not By elif condition
    
    print("The weather is warm.")          # elif Condition is True Then print  The weather is warm.
else:                                      # Else: If/elif Condition is Flase  then else block execute
    print("It's a bit cool.")              # Then print It's a bit cool.
```

# 13. Functions 

```python
def analyze_sentiment(score):             # def function header, analyze_sentiment-Function name- score- parameter
    if score > 0:                         # Checking condition with if statement.
        return "Positive"                 # If condition satisfy then return "Positive"
    else:                                 # Else condition Not satisfy then return "Negative" 
        return "Negative" 
    print(analyze_sentiment(0.8))         # Print analyze_sentiment
```

# 15. Using Pandas DataFrames 

```python
data = {'feature1': [1, 2, 3, 4, 5], 'feature2': [10, 20, 30, 40, 50]} 
df = pd.DataFrame(data) 
print("DataFrame created:") 
print(df) 
print("\nFirst 3 rows of the DataFrame:") 
print(df.head(3)) 
print("\nBasic statistics of the DataFrame:") 
print(df.describe())
```

```python
#data is a Python dictionary.It contains two keys: feature1 → [1, 2, 3, 4, 5] feature2 → [10, 20, 30, 40, 50]
# You can think of each key as a column name.
# pd.DataFrame() converts the dictionary into a Pandas DataFrame.
# Print data frame
# df.head print "Give me the first 3 rows of the DataFrame."
# df.describe() calculates statistics for the numerical columns like mean,count, std, min, max,25%, 50%, 75% max.
```
