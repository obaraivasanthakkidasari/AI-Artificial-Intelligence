Program to add "n" natural numbers ---> sum= 1+2+3+4.....+n

```python
n = int(input("Enter n:"))
sum =0
i = 1

while i<=n:
    sum = sum+i
    i = i+1
    
print(sum)
```

Program to find products of  all numbers present in a list

```python
lst = [1,2,3,4,6]

p=1
i=0

while i <len(lst):
    p = p*lst[i]
    i=i+1
    
print("Product is:",p)
```

# Infinite Loop

```python
print("Enter your values and once everything is completed enter #")

s=0
      
while True:
      a= input('enter a value:')
      if(a=='#'):
          break
      s=s+int(a)
      
print(s)
```

# While loop with else

```python
numbers = [1,2,3,4,5]

index =0

while index <= len(numbers):
    print(index)
    index = index+1
    if index==3:
        break
        
else:
    print("No item left in the list")
```
